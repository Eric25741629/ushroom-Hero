```python

from tools import *
import easyocr
from mask import *
import cv2
import numpy as np
import uiautomator2 as u2
import time

def click_white(d: u2.Device):
    d.click(509, 56)
    time.sleep(1)
# 計算停了幾台車


def swipe(d: u2.Device):
    d.touch.down(0.5, 0.6)  # 模拟按下
    time.sleep(.01)  # down 和 move 之间的延迟，自己控制
    for i in range(40):
        d.touch.move(0.5, 0.6 - i * 0.01)
    time.sleep(0.3)  # down 和 move 之间的延迟，自己控制
    d.touch.up(0.5, 0.6 - i * 0.01)  # 模拟抬起


def count_car(img):
    img = img[138:259]
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    lower = np.array([19, 26, 191])
    upper = np.array([34, 73, 255])
    mask = cv2.inRange(hsv, lower, upper)
    mask = cv2.dilate(mask, None, iterations=2)
    mask = cv2.erode(mask, None, iterations=2)
    contours, _ = cv2.findContours(
        mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    num_boxes = 0
    for contour in contours:
        area = cv2.contourArea(contour)
        if area > 1500:
            num_boxes += 1
    return num_boxes


def check_world_car(img):
    img = img[138:259, 45:145]
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    lower = np.array([19, 50, 229])
    upper = np.array([45, 156, 255])
    mask = cv2.inRange(hsv, lower, upper)
    mask = cv2.dilate(mask, None, iterations=2)
    mask = cv2.erode(mask, None, iterations=2)
    contours, _ = cv2.findContours(
        mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    num_boxes = 0
    for contour in contours:
        area = cv2.contourArea(contour)
        if area > 1500:
            num_boxes += 1
    if num_boxes > 0:
        return True
    return False


def swipe_car(d):
    d.touch.down(186, 670)  # 模拟按下
    time.sleep(.01)  # down 和 move 之间的延迟，自己控制
    for i in range(39):
        d.touch.move(186-i*2, 670)
    time.sleep(0.5)  # down 和 move 之间的延迟，自己控制
    d.touch.up(186-i*2, 670)  # 模拟抬起
    time.sleep(1)
    d.click(70, 670)


def check_start_button(img: np.ndarray = None):
    if img is None:
        img = cv2.imread(r'A:\Screenshots\2.png')
    hsv = cv2.cvtColor(img[741:819,], cv2.COLOR_BGR2HSV)
    lower = green_mask_lower
    upper = green_mask_upper
    mask = cv2.inRange(hsv, lower, upper)
    if np.sum(mask) > 150:
        return True
    return False


def full_park(img: np.ndarray = None):
    if img is None:
        img = cv2.imread(r'A:\Screenshots\2.png')
    hsv = cv2.cvtColor(img[595:622,], cv2.COLOR_BGR2HSV)
    lower = hr_mask_lower
    upper = hr_mask_upper
    mask = cv2.inRange(hsv, lower, upper)
    if np.sum(mask) > 150:
        return True
    return False


def get_park_xy(d):
    img = d.screenshot(format='opencv')
    img1 = cv2.imread('park.jpg')
    # 找圖
    res = cv2.matchTemplate(img, img1, cv2.TM_CCOEFF_NORMED)
    loc = np.where(res >= 0.8)
    loc = non_max_suppression(np.array(loc).T, 10)
    print(len(loc))
    # 繪至方框在圖片上
    can_park = []
    for pt in loc:
        cv2.rectangle(img, (pt[1], pt[0]), (pt[1] + img1.shape[1],
                      pt[0] + img1.shape[0]), (0, 0, 255), 2)
        img2 = img[pt[0]:pt[0] + img1.shape[0], pt[1]:pt[1] + img1.shape[1]]
        hsv = cv2.cvtColor(img2, cv2.COLOR_BGR2HSV)
        lower = np.array([26, 95, 167])
        upper = np.array([85, 208, 255])
        mask = cv2.inRange(hsv, lower, upper)
        contours, _ = cv2.findContours(
            mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        num_boxes = 0
        for contour in contours:
            area = cv2.contourArea(contour)
            if area > 55:
                num_boxes += 1
                break
        if num_boxes == 0:
            pass
        else:
            print("停了")
            continue
        img2 = img[pt[0]-10:pt[0] + img1.shape[0], :119]
        full_img = cv2.imread('full.jpg')
        res = cv2.matchTemplate(img2, full_img, cv2.TM_CCOEFF_NORMED)
        loc = np.where(res >= 0.8)
        if len(loc[0]) > 0:
            print("滿了")
            continue
        can_park.append(pt)
    return can_park

def find_place(img):
    # 轉到hsv
    y=[208,433]
    x=[84,389]
    place =[]
    place_num =1
    for i in y:
        # 分成四塊
        for j in x:
            time_place = img[i:i+28,j:j+100]
            hsv = cv2.cvtColor(time_place, cv2.COLOR_BGR2HSV)
            # 設定顏色範圍
            lower = np.array([6, 0, 85])
            higher = np.array([41,139,108])
            # 設定mask
            mask = cv2.inRange(hsv, lower, higher)
            # 膨脹
            mask = cv2.erode(mask, None, iterations=1)
            mask = cv2.dilate(mask, None, iterations=30)
            # 腐蝕
            if(np.sum(mask))<352000:
                place.append(place_num)
            place_num += 1
    return place
def Collapse(img):
    # 尋找圖片
    img = img[180:218,514:]
    img1 = cv2.imread('collapse.jpg')
    res = cv2.matchTemplate(img, img1, cv2.TM_CCOEFF_NORMED)
    loc = np.where(res >= 0.8)
    loc = non_max_suppression(np.array(loc).T, 10)
    # 繪至方框在圖片上
    if len(loc):
        return True
    return False
def park(d: u2.Device, reader):
    d.click(321, 913)
    time.sleep(3)
    d.click(451, 451)
    time.sleep(2)
    d.click(29, 204)
    time.sleep(2)
    d.click(368, 515)
    time.sleep(2)
    click_white(d)
    # world_car = check_world_car(d.screenshot(format='opencv'))
    d.swipe(300, 200, 100, 200)
    time.sleep(4)
    car_num = count_car(d.screenshot(format='opencv'))
    print(car_num)
    if 5-car_num == 0:
        print("停車位已滿")
        d.click(475,919)
        time.sleep(2)
        d.click(321, 913)
        return
    else:
        d.click(275, 913)
        time.sleep(2)
        while (5-car_num):
            # car_num =count_car(d.screenshot(format='opencv'))
            park_list = get_park_xy(d)
            if len(park_list) == 0:
                print("沒有停車位")
                swipe(d)
                time.sleep(1)
                continue
            else:
                for i in park_list:
                    print(i)
                    d.click(int(i[1])+30, int(i[0])+10)
                    time.sleep(1)
                    parkpoint = [[124, 316], [430, 316], [124, 536], [430, 536]]
                    if Collapse(d.screenshot(format='opencv')):
                        d.click(528,202)
                        time.sleep(1)
                    can_park = find_place(d.screenshot(format='opencv'))
                    if len(can_park) == 0:
                        break;
                    parkpoint = [parkpoint[i-1] for i in can_park]
                    for j in parkpoint:
                        d.click(j[0], j[1])
                        time.sleep(2)
                        img = d.screenshot(format='opencv')
                        result = reader.readtext(img, detail=0)
                        print(result)
                        if '奇遇獎勵預覽' not in result and '今日' not in str(result):
                            click_white(d)
                            time.sleep(1)
                            continue
                        else:
                            # 如果綠色開始停車 和未滿12小時 那就停車
                            error = 0
                            while (1):
                                print("找開始停車")
                                img = d.screenshot(format='opencv')
                                if check_start_button(img) and not full_park(img):
                                    d.click(272, 792)
                                    time.sleep(1)
                                    img = d.screenshot(format='opencv')

                                    result = reader.readtext(img, detail=0)
                                    if "取消" in result:
                                        d.click(375, 554)
                                    else:
                                        d.click(277, 616)
                                    break
                                else:
                                    print("繼續找")
                                    swipe_car(d)
                                    time.sleep(1)
                                    error += 1
                                    if error > 10:
                                        break
                            img = d.screenshot(format='opencv')
                            hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
                            lower = green_mask_lower
                            upper = green_mask_upper
                            mask = cv2.inRange(hsv, lower, upper)[590:641,]
                            if np.sum(mask) > 150:
                                click_white(d)

                            # if car_num == 0:
                            #     break
                            # if np.sum(mask) > 150:
                            #     click_white(d)
                            #     time.sleep(1)
                            #     d.click(273, 892)
                            # d.swipe(300, 200, 100, 200)
                            # time.sleep(2)
                            # car_num = count_car(d.screenshot(format='opencv'))
                            # time.sleep(3)
                            # print(car_num)
            if car_num == 0:
                break

            # d.click(i[1],i[0])



reader = easyocr.Reader(['ch_sim', 'en'])
```
