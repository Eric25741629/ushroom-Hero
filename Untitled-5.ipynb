```python
# 系统指令 + 用户输入
system_instruction = "你是一个智能助手，负责生成控制设备的 MQTT 消息，请严格按照以下 JSON 格式输出，输出内容无任何多余解释：範例: {\"device\": \"light\", \"id\": 1, \"action\": \"on\"}"
example_format = '{"device": "light", "id": 1, "action": "on"}'
user_query = "幫我把一號燈關閉"

input_text=system_instruction + user_query
response = pipe(
    input_text,
    max_new_tokens=20,
    temperature=0.1,  # 减少随机性
    top_p=0.8,       # 限制输出内容集中于高概率的词
    repetition_penalty=1.2,  # 惩罚重复
)
print(response[0]["generated_text"].replace(system_instruction+ user_query, "").strip())
```

```python
# 拼接提示词
user_query = "请帮我打開2號電燈。"
input_prompt = f"{system_instruction}{example_format}\n\n用户输入: {user_query}\nMQTT消息:"

# 使用模型生成结果
response = pipe(
    input_prompt,
    max_new_tokens=50,  # 控制生成长度
    temperature=0.1,    # 降低随机性，确保格式一致
    top_p=0.9,
)

# 提取生成文本
mqtt_message = response[0]["generated_text"].replace(input_prompt, "").strip()
print(f"Generated MQTT Message: {mqtt_message}")
```

```python
import re
import json
response[0]["generated_text"]
def encode_mqtt_message(str1):

    # 匹配 JSON 格式内容
    json_match = re.search(r"{.*}", str1)
    if json_match:
        mqtt_message = json.loads(json_match.group(0))  # 转换为 JSON 对象
        print("生成的 MQTT 消息：", mqtt_message)
    else:
        print("未找到有效 JSON 格式")
encode_mqtt_message(response[0]["generated_text"])
```
