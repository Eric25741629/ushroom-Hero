# spin_and_send_gold single test

This notebook keeps Jupyter for convenience, but runs Playwright in a separate Python process to avoid the asyncio loop conflict.

```python
from pathlib import Path
import os
import subprocess
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "bot_config.json").exists():
    PROJECT_ROOT = Path(r"A:\菇勇者全自動掛機")
os.chdir(PROJECT_ROOT)

RUNNER = PROJECT_ROOT / "spin_and_send_gold_single_runner.py"
assert RUNNER.exists(), RUNNER

print("PROJECT_ROOT =", PROJECT_ROOT)
print("RUNNER =", RUNNER)
```

```python
TARGET_IP = "emulator-5558"
SCREENSHOT_PATH = PROJECT_ROOT / "debug_img" / f"{TARGET_IP}_spin_test.png"

TARGET_IP
```

```python
def run_cmd(args):
    cmd = [sys.executable, str(RUNNER), *args]
    print("CMD:", " ".join(cmd))
    cp = subprocess.run(cmd, cwd=PROJECT_ROOT, text=True, capture_output=True)
    if cp.stdout:
        print(cp.stdout)
    if cp.returncode != 0:
        if cp.stderr:
            print(cp.stderr)
        raise RuntimeError(f"command failed: {cp.returncode}")
    return cp
```

## Observe Only

If you only want to watch the browser, run the next cell. Interrupt the cell to close the browser.

```python
run_cmd(["observe", "--ip", TARGET_IP])
```
